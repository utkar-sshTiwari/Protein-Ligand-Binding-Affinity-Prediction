```python
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torch.utils.tensorboard import SummaryWriter

# keep only cells that actually have a value, in row-major order
rows, cols = np.where(~np.isnan(pKd))
labels = pKd[rows, cols]
print("total pairs:", len(labels))          # expect 30056

# official split: indices into the arrays above
train_folds = json.load(open(root + "folds/train_fold_setting1.txt"))
test_idx    = json.load(open(root + "folds/test_fold_setting1.txt"))

val_idx   = train_folds[0]                                   # fold 0 = validation
train_idx = [i for f in train_folds[1:] for i in f]          # folds 1-4 = training
print("train:", len(train_idx), "val:", len(val_idx), "test:", len(test_idx))
```

```python
class AffinityDataset(Dataset):
    def __init__(self, idx):
        self.idx = idx

    def __len__(self):
        return len(self.idx)

    def __getitem__(self, i):
        k = self.idx[i]
        drug    = torch.tensor(drug_enc[rows[k]],    dtype=torch.long)
        protein = torch.tensor(protein_enc[cols[k]], dtype=torch.long)
        y       = torch.tensor(labels[k],            dtype=torch.float)
        return drug, protein, y

BATCH_SIZE = 256
train_loader = DataLoader(AffinityDataset(train_idx), batch_size=BATCH_SIZE, shuffle=True)
val_loader   = DataLoader(AffinityDataset(val_idx),   batch_size=BATCH_SIZE)
test_loader  = DataLoader(AffinityDataset(test_idx),  batch_size=BATCH_SIZE)

d, p, y = next(iter(train_loader))
print(d.shape, p.shape, y.shape)   # expect [256, 85] [256, 1200] [256]
```

```python
class CNNBranch(nn.Module):
    """Embedding -> 3 conv layers -> global max pool. One of these per input."""
    def __init__(self, vocab_size, kernel, embed_dim=128, filters=32):
        super().__init__()
        self.embed = nn.Embedding(vocab_size + 1, embed_dim, padding_idx=0)  # +1 for padding id 0
        self.convs = nn.Sequential(
            nn.Conv1d(embed_dim,    filters,     kernel), nn.ReLU(),
            nn.Conv1d(filters,      filters * 2, kernel), nn.ReLU(),
            nn.Conv1d(filters * 2,  filters * 3, kernel), nn.ReLU(),
        )

    def forward(self, x):                 # x: [B, L] integer ids
        x = self.embed(x)                 # [B, L, E]
        x = x.transpose(1, 2)             # [B, E, L]  (Conv1d wants channels first)
        x = self.convs(x)                 # [B, 96, L']
        return torch.max(x, dim=2).values # [B, 96]  global max pool over the sequence


class DeepDTA(nn.Module):
    def __init__(self, smiles_vocab_size, protein_vocab_size):
        super().__init__()
        self.drug_branch    = CNNBranch(smiles_vocab_size,  kernel=4)
        self.protein_branch = CNNBranch(protein_vocab_size, kernel=8)
        self.head = nn.Sequential(
            nn.Linear(96 + 96, 1024), nn.ReLU(), nn.Dropout(0.1),
            nn.Linear(1024, 1024),    nn.ReLU(), nn.Dropout(0.1),
            nn.Linear(1024, 512),     nn.ReLU(),
            nn.Linear(512, 1),
        )

    def forward(self, drug, protein):
        d = self.drug_branch(drug)
        p = self.protein_branch(protein)
        return self.head(torch.cat([d, p], dim=1)).squeeze(1)  # [B]

model = DeepDTA(len(smiles_vocab), len(protein_vocab))
print(sum(p.numel() for p in model.parameters()), "parameters")
print(model(d, p).shape)    # expect torch.Size([256]), reusing the batch from Cell 2
```

```python
def concordance_index(y_true, y_pred):
    """Of all pairs with different true values, the fraction where the model ranks them correctly."""
    y_true, y_pred = np.asarray(y_true), np.asarray(y_pred)
    num, den = 0.0, 0.0
    for i in range(len(y_true)):
        mask = y_true[i] > y_true
        den += mask.sum()
        diff = y_pred[i] - y_pred[mask]
        num += (diff > 0).sum() + 0.5 * (diff == 0).sum()
    return num / den

@torch.no_grad()
def evaluate(loader):
    model.eval()
    preds, trues = [], []
    for drug, protein, y in loader:
        preds.append(model(drug, protein))
        trues.append(y)
    preds = torch.cat(preds).numpy()
    trues = torch.cat(trues).numpy()
    mse = float(np.mean((preds - trues) ** 2))
    return mse, preds, trues
```

```python
EPOCHS = 30        # the paper used 100; raise this once you know the speed
LR = 1e-3

optimizer = torch.optim.Adam(model.parameters(), lr=LR)
loss_fn = nn.MSELoss()
writer = SummaryWriter("runs/deepdta_davis")

best_val = float("inf")
for epoch in range(EPOCHS):
    model.train()
    total = 0.0
    for drug, protein, y in train_loader:
        optimizer.zero_grad()
        loss = loss_fn(model(drug, protein), y)
        loss.backward()
        optimizer.step()
        total += loss.item() * len(y)
    train_mse = total / len(train_idx)

    val_mse, _, _ = evaluate(val_loader)
    writer.add_scalar("MSE/train", train_mse, epoch)
    writer.add_scalar("MSE/val", val_mse, epoch)
    print(f"epoch {epoch+1:3d} | train MSE {train_mse:.4f} | val MSE {val_mse:.4f}")

    if val_mse < best_val:
        best_val = val_mse
        torch.save(model.state_dict(), "deepdta_davis_best.pt")
```

```python
model.load_state_dict(torch.load("deepdta_davis_best.pt"))
test_mse, preds, trues = evaluate(test_loader)
test_ci = concordance_index(trues, preds)
print(f"Test MSE: {test_mse:.3f} | Test CI: {test_ci:.3f}")
```
